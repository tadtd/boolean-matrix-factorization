### Load Chess dataset

```python
from ucimlrepo import fetch_ucirepo 
from sklearn.preprocessing import OneHotEncoder
  
# fetch dataset 
chess_king_rook_vs_king_pawn = fetch_ucirepo(id=22) 
  
data = chess_king_rook_vs_king_pawn.data.features 

encoder = OneHotEncoder(sparse_output=False, dtype=int)
X = encoder.fit_transform(data)
X = X.astype(int)

print("Chess dataset shape:", X.shape)
print("Number of 1s:", X.sum())
```

### Asso algorithm

```python
from BMF import Asso

model = Asso(rank=90)
result = model.solve(X)
print(result)
```
