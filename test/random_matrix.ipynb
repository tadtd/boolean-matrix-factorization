### Create random binary matrix

```python
from BMF import utils
m = 70
n = 50
density = 0.3
A = utils.random_boolean_matrix(m, n, density=density, random_state=42)
```

### Asso algorithm

```python
from BMF import Asso

model = Asso(rank=50)
result = model.solve(A)
print(result)
```

### GreCon algorithm

```python
from BMF import GreCon
model = GreCon()
result = model.solve(A)
print(result)
```

### GreConD algorithm

```python
from BMF import GreConD
model = GreConD()
result = model.solve(A)
print(result)
```

### GreConD+ algorithm

```python
from BMF import GreConDPlus

w = 4
model = GreConDPlus(w=w)
result = model.solve(A)
print(result)
```
