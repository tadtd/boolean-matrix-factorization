### Create random binary matrix

```python
from BMF import utils
m = 70
n = 50
density = 0.3
A = utils.random_boolean_matrix(m, n, density=density, random_state=42)
```

### Asso algorithm

```python
from BMF import Asso

model = Asso(rank=50)
result = model.solve(A)
print(result)
```

### GreCon algorithm

```python
from BMF import GreCon
model = GreCon()
result = model.solve(A)
print(result)
```

### GreConD algorithm

```python
from BMF import GreConD
model = GreConD()
result = model.solve(A)
print(result)
```

### GreConD+ algorithm

```python
from BMF import GreConDPlus

w = 4
model = GreConDPlus(w=w)
result = model.solve(A)
print(result)
```

```python
import numpy as np

from BMF import Asso, GreCon, GreConD, GreConDPlus

densities = [0.2, 0.3, 0.4]
w_vals = [2, 4, 6]

for density in densities:
    A = utils.random_boolean_matrix(100, 100, density=density, random_state=42)
    print(f"\n--- Density = {density} ---")
    print(f"Total 1s: {np.sum(A)}")
    
    # Asso
    print("Running Asso...")
    model = Asso(rank=80)
    result = model.solve(A)
    print(f"Asso: Rank={result.rank}, Ratio={result.coverage_equality:.4f}, Error={result.error:.0f}, Time={result.time_taken:.2f}s")
    
    print('Running GreCon...')
    model = GreCon()
    result = model.solve(A)
    print(f"GreCon: Rank={result.rank}, Ratio={result.coverage_equality:.4f}, Error={result.error:.0f}, Time={result.time_taken:.2f}s")

    # GreConD
    print("Running GreConD...")
    model = GreConD()
    result = model.solve(A)
    print(f"GreConD: Rank={result.rank}, Ratio={result.coverage_equality:.4f}, Error={result.error:.0f}, Time={result.time_taken:.2f}s")
    
    # GreConD+ with different w values
    for w in w_vals:
        print(f"Running GreConD+ (w={w})...")
        model = GreConDPlus(w=w)
        result = model.solve(A)
        print(f"GreConD+ (w={w}): Rank={result.rank}, Ratio={result.coverage_equality:.4f}, Error={result.error:.0f}, Time={result.time_taken:.2f}s")
```
